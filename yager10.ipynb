```python
import math

lam = 0.7
p = 7

print("x\tmu\tC\tugeno\tYager")

for x in range(7):

    # Gaussian membership function
    mu = math.exp(-(x*x)/8)

    # Standard complement
    c = 1 - mu

    # Sugeno complement
    s = (1-mu)/(1+lam*mu)

    # Yager complement
    y = (1-mu**p)**(1/p)

    print(x, "\t", round(mu,3), "\t",
          round(c,3), "\t",
          round(s,3), "\t",
          round(y,3))
```
