```python
p1 = "11001100"
p2 = "00110011"
print("p1:", p1)
print("p2:", p2)
# 1 point crossover
point = 4
child1 = p1[:point]+p2[point:]
child2 = p2[:point]+p1[point:]
print("child1=",child1)
print("child2=",child2)
#2 point crossover
p = 2
q = 6
child1 = p1[:p]+p2[p:q]+p1[q:]
child2 = p2[:p]+p1[p:q]+p2[q:]
print("2 point crossover:")
print("child1=",child1)
print("child2=",child2)
#uniform crossover
mask ="10101010"
c1 = ""
c2 = ""
for i in range(len(mask)):
    if mask[i] == "1":
        c1 = c1 +p1[i]
        c2 = c2 +p2[i]
    else:
        c1 = c1 +p2[i]
        c2 = c2 +p1[i]
print("uniform crossover:")
print("child1=",c1)
print("child2=",c2)
```
