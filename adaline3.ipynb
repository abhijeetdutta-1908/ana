```python
x1 = [-1, -1, 1, 1]
x2 = [-1, 1, -1, 1]
t = [-1, -1, -1, 1]

w1 = 0.1
w2 = 0.1
b = 0.1
eta = 0.1

print("x1\tx2\tx0\tt\tyin\te\tdw1\tdw2\tdb\tw1\tw2\tb")

for epoch in range(10):
    print("\nEpoch:", epoch+1)

    for i in range(4):
        x0 = 1

        yin = b + x1[i]*w1 + x2[i]*w2

        e = t[i] - yin

        dw1 = eta * e * x1[i]
        dw2 = eta * e * x2[i]
        db = eta * e

        w1 = w1 + dw1
        w2 = w2 + dw2
        b = b + db

        print(x1[i], "\t", x2[i], "\t", x0, "\t", t[i], "\t", round(yin, 2), "\t", round(e, 2), "\t", round(dw1, 2), "\t", round(dw2, 2), "\t", round(db, 2), "\t", round(w1, 2), "\t", round(w2, 2), "\t", round(b, 2))

print("\nFinal weights:", w1, w2, b)
```
