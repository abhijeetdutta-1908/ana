Hebbian learning

```python
import math
x1 = 2
x2 = 1
w1 = 0.5
w2 = 0.2
eta = 0.1 
print("x1\tx2\ty\tdw1\tdw2\tw1\tw2\tnorm")
for i in range(1,16):
    yin = x1*w1 + x2*w2

    dw1 = eta*x1*yin
    dw2 = eta*x2*yin

    w1 = w1 + eta*x1*yin
    w2 = w2 + eta*x2*yin
    norm = math.sqrt(w1**2 + w2**2)
    print(x1,"\t", x2,"\t",round(yin,1),"\t",round(dw1,1),"\t",round(dw2,1),"\t",round(w1,1),"\t",round(w2,1),"\t",round(norm,1))
    

```
