```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.linspace(0, 10, 100)

# Create 2D grid
X, Y = np.meshgrid(x, y)

A = np.exp(-(((X-5)**2)/(2*1.5**2) +
             ((Y-5)**2)/(2*2.0**2)))

A1 = np.max(A, axis=0)

plt.plot(x, A1)

plt.xlabel("x")
plt.ylabel("Membership Value")
plt.title("Projection of 2D Gaussian to 1D")
plt.grid(True)
plt.show()
```
