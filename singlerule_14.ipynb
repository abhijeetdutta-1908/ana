```python
import numpy as np

# Fuzzy sets
A = np.array([0, 0.8, 0.6, 1])
A_prime = np.array([0.2, 0.7, 0.5, 0.9])  # Assumed values

B = np.array([0.2, 1, 0.8, 0])
C = np.array([0, 0.4, 1, 0.8])

# Calculate matching degree
alpha = np.max(np.minimum(A, A_prime))

# Fuzzy evaluation
B_prime = np.maximum(
    np.minimum(alpha, B),
    np.minimum(1 - alpha, C)
)

print("Matching degree =", alpha)
print("Final fuzzy set B' =", B_prime)
```
