```python
import numpy as np
import matplotlib.pyplot as plt

a, b, c = 20, 60, 80
core_tri = b
cross_tri = [a + (b-a)*0.5, b + (c-b)*0.5]
bw_tri = cross_tri[1] - cross_tri[0]

a, b, c, d = 10, 20, 40, 70
core_trap = [b, c]
cross_trap = [a + (b-a)*0.5, c + (d-c)*0.5]
bw_trap = cross_trap[1] - cross_trap[0]

mean, sigma = 5, 1.5
core_gauss = mean
cross_gauss = [
    mean - sigma*np.sqrt(2*np.log(2)),
    mean + sigma*np.sqrt(2*np.log(2))
]
bw_gauss = cross_gauss[1] - cross_gauss[0]

a, b, c = 1.5, 2, 5
core_bell = c
cross_bell = [c-a, c+a]
bw_bell = cross_bell[1] - cross_bell[0]

print("Triangular")
print("Core:", core_tri)
print("Crossover:", cross_tri)
print("Bandwidth:", bw_tri)

print("\nTrapezoidal")
print("Core:", core_trap)
print("Crossover:", cross_trap)
print("Bandwidth:", bw_trap)

print("\nGaussian")
print("Core:", core_gauss)
print("Crossover:", cross_gauss)
print("Bandwidth:", round(bw_gauss, 2))

print("\nBell")
print("Core:", core_bell)
print("Crossover:", cross_bell)
print("Bandwidth:", bw_bell)

x1 = np.linspace(0, 100, 500)
x2 = np.linspace(-2, 12, 500)

tri = np.maximum(0, np.minimum((x1-20)/40, (80-x1)/20))
trap = np.maximum(0, np.minimum(np.minimum((x1-10)/10, 1), (70-x1)/30))
gauss = np.exp(-((x2-5)**2)/(2*1.5**2))
bell = 1/(1+abs((x2-5)/1.5)**4)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(x1, tri)
plt.title("Triangular")
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(x1, trap)
plt.title("Trapezoidal")
plt.grid()

plt.subplot(2, 2, 3)
plt.plot(x2, gauss)
plt.title("Gaussian")
plt.grid()

plt.subplot(2, 2, 4)
plt.plot(x2, bell)
plt.title("Bell")
plt.grid()

plt.tight_layout()
plt.show()
```
