```python
w1 = 1
w2 = 1
def neuron(x1,x2,t):
    y = w1*x1 + w2*x2
    if y >= t:
        return 1
    else:
        return 0
print("x1\tx2\tand\tor\tnand")
for x1 , x2 in[(0,0),(0,1),(1,0),(1,1)]:
    a= neuron(x1,x2,2)
    o = neuron(x1,x2,1)
    nand = neuron(-x1,-x2,-1)
    print(x1, "\t",x2, "\t",a, "\t",o, "\t",nand)
```
