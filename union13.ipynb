```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11)
A = x / (x + 1)
B = 2 / (x + 2)

union = np.maximum(A, B)
intersection = np.minimum(A, B)

complement_union = 1 - union
complement_intersection = 1 - intersection

# Plot
plt.plot(x, A, label="A")
plt.plot(x, B, label="B")
plt.plot(x, union, label="Union")
plt.plot(x, intersection, label="Intersection")
plt.plot(x, complement_union, label="Complement of Union")
plt.plot(x, complement_intersection, label="Complement of Intersection")

plt.xlabel("x")
plt.ylabel("Membership Value")
plt.title("Fuzzy Set Operations")
plt.legend()
plt.grid(True)
plt.show()
```
