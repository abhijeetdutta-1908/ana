Max-Min Composition

```python
import math

x = [1, 2, 3]
y = [1, 2, 3]
z = [1, 2, 3]

R1 = []
R2 = []

for i in x:
    row = []
    for j in y:
        a = math.exp(-((i-j)**2)/(2*1.5**2))
        row.append(round(a, 2))
    R1.append(row)

for i in y:
    row = []
    for j in z:
        a = math.exp(-((i-j)**2)/(2*2**2))
        row.append(round(a, 2))
    R2.append(row)

print("R1:")
for row in R1:
    print(row)

print("R2:")
for row in R2:
    print(row)

R = []

for i in range(len(x)):
    row = []
    for j in range(len(z)):
        values = []
        for k in range(len(y)):
            values.append(min(R1[i][k], R2[k][j]))
        row.append(round(max(values), 2))
    R.append(row)

print("Max-Min Composition:")
for row in R:
    print(row)
```
